```python
import asyncio
from pathlib import Path

import pygame

# Stop the previous monitor if this cell is rerun.
if "intro_monitor_task" in globals():
    intro_monitor_task.cancel()

# Local audio folders
audio_dir = Path(r"D:\KAOS\visionNov132026-EDITS\audio")
tones_dir = audio_dir / "phoneTones"
recordings_dir = audio_dir / "weLoveLeimertArchive-Nov132026"

pygame.mixer.init()
pygame.mixer.stop()

# Keypad-to-recording mappings
recordings = {
    "0": "noPriorArt1.wav",
    "1": "AmaraKiona.wav",
    "2": "BenCaldwell.wav",
    "3": "Benu.wav",
    "4": "HughAugustine.wav",
    "5": "Jimetta.wav",
    "6": "MaryChristian.wav",
    "7": "MattGibson.wav",
    "8": "Mitebrafoagindotan.wav",
    "9": "WeLoveLeimertAudioCollage.wav",
}

intro_sound = pygame.mixer.Sound(str(recordings_dir / recordings["0"]))
phone_off_hook = False


def pick_up():
    global phone_off_hook

    if phone_off_hook:
        print("Phone is already picked up.")
        return

    pygame.mixer.stop()
    phone_off_hook = True
    pygame.mixer.Channel(0).play(intro_sound, loops=-1)
    print(f"Phone picked up. Playing intro: {recordings['0']}")


def hang_up():
    global phone_off_hook

    phone_off_hook = False
    pygame.mixer.stop()
    print("Phone hung up. Audio stopped.")


def play_selection(selected_key):
    if not phone_off_hook:
        print("Pick up the phone before selecting a recording.")
        return

    filename = recordings[selected_key]
    tone = pygame.mixer.Sound(str(tones_dir / f"DTMF-{selected_key}.wav"))
    recording = pygame.mixer.Sound(str(recordings_dir / filename))

    pygame.mixer.stop()

    channel = pygame.mixer.Channel(0)
    channel.play(tone)
    channel.queue(recording)

    print(f"Selected key {selected_key}: {filename}")


async def monitor_playback():
    while True:
        await asyncio.sleep(0.1)

        channel = pygame.mixer.Channel(0)

        if phone_off_hook and not channel.get_busy():
            channel.play(intro_sound, loops=-1)
            print("Recording finished. Returning to intro.")


intro_monitor_task = asyncio.create_task(monitor_playback())

print("Prototype ready. Phone is hung up.")
```
